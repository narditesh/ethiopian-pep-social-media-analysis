```python
df['role'].value_counts()
```

```python
def categorize_role(role):
    role = str(role)

    if "Minister" in role:
        return "Minister"

    elif "President" in role:
        return "President"

    elif "Prime Minister" in role:
        return "Prime Minister"

    elif "Speaker" in role:
        return "Parliament Leadership"

    elif "Mayor" in role:
        return "Mayor"

    else:
        return "Other"

df['role_category'] = df['role'].apply(categorize_role)
```

```python
df['role_category'].value_counts()
```

```python
import matplotlib.pyplot as plt

df['role_category'].value_counts().plot(kind='bar')
plt.title('PEPs by Role Category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

```python
social_cols = [
    'fb_url',
    'instagram_url',
    'threads_url',
    'x_url',
    'tiktok_url',
    'youtube_url',
    'linkedin_url',
    'website_url',
    'wikipedia_url',
    'telegram'
]

social_presence = {}

for col in social_cols:
    social_presence[col] = (df[col] != 'N/A').sum()

print(social_presence)
```

```python
for col in social_cols:
    print(f"\n{col}")
    print(df[col].value_counts().head())
```

```python
import numpy as np

df = df.replace(['N/A', 'N/A.1', 'N/A.2', 'N/A.3', 'N/A.4', 'N/A.5', 'N/A.6'], np.nan)
```

```python
social_presence = {}

for col in social_cols:
    social_presence[col] = df[col].notna().sum()

print(social_presence)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

pd.Series(social_presence).sort_values(ascending=False).plot(kind='bar')

plt.title('Digital Presence of Ethiopian PEPs')
plt.xlabel('Platform')
plt.ylabel('Number of Profiles')
plt.xticks(rotation=45)
plt.show()
```

```python
social_presence_series = pd.Series(social_presence)
social_presence_series.sort_values(ascending=False)
```
